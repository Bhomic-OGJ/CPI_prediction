```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set Seaborn style for a beautiful look
sns.set_style("whitegrid")
sns.set_context("talk", font_scale=1.2, rc={"font.family": "Helvetica", "lines.linewidth": 2.5})

data = res.copy()
# Data from the provided table
# data = {
#     'Epoch': [0, 1, 2, 3],
#     'Time (sec)': [19.470728, 40.056900, 61.370864, 84.042677],
#     'Loss_train': [829.656900, 670.824770, 580.226900, 528.704830],
#     'AUC_dev': [0.888610, 0.915789, 0.923331, 0.930086],
#     'AUC_test': [0.771520, 0.848748, 0.869048, 0.884768],
#     'Precision_test': [0.854545, 0.670000, 0.881356, 0.773333],
#     'Recall_test': [0.602564, 0.858974, 0.666667, 0.743590]
# }

# Create DataFrame
df = pd.DataFrame(data)

# Create figure and axis objects with a secondary y-axis
fig, ax1 = plt.subplots(figsize=(12, 7))

# Define a vibrant color palette
colors = sns.color_palette("husl", 5)  # 5 colors for the 5 metrics

# Plot performance metrics on primary y-axis
ax1.plot(df['Epoch'], df['Loss_train'], marker='o', label='Loss (Train)', color=colors[0], markersize=10)
# ax1.plot(df['Epoch'], df['AUC_dev'] * 1000, marker='s', label='AUC (Dev) × 1000', color=colors[1], markersize=10)
# ax1.plot(df['Epoch'], df['AUC_test'] * 1000, marker='^', label='AUC (Test) × 1000', color=colors[2], markersize=10)
# ax1.plot(df['Epoch'], df['Precision_test'] * 1000, marker='d', label='Precision (Test) × 1000', color=colors[3], markersize=10)
# ax1.plot(df['Epoch'], df['Recall_test'] * 1000, marker='*', label='Recall (Test) × 1000', color=colors[4], markersize=10)

# Set labels and title for primary y-axis
ax1.set_xlabel('Epoch', fontsize=14, weight='bold')
ax1.set_ylabel('Loss', fontsize=14, weight='bold')
# ax1.set_title('Performance Metrics Over Epochs for QEFS', fontsize=16, weight='bold', pad=20)

# Create secondary y-axis for Time
# ax2 = ax1.twinx()
# ax2.plot(df['Epoch'], df['Time(sec)'], marker='x', label='Time (sec)', color='teal', linestyle='--', markersize=10)
# ax2.set_ylabel('Time (sec)', fontsize=14, weight='bold', color='teal')
# ax2.tick_params(axis='y', labelcolor='teal')

# Combine legends from both axes
# lines1, labels1 = ax1.get_legend_handles_labels()
# lines2, labels2 = ax2.get_legend_handles_labels()
# ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3, fontsize=10, frameon=True, edgecolor='black')

# Adjust layout to prevent overlap
plt.tight_layout()

# Save the plot as a high-resolution PNG
plt.savefig('Loss_train.png', dpi=300, bbox_inches='tight')
plt.close()
```
